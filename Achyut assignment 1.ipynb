```python
#8. write a program which takes numerator and denominator values from the user and gives output as tuple of quotient and remainder

numerator = int(input("enter numerator:"))
denominator = int(input("enter denominator:"))
print(numerator%denominator)
print(numerator//denominator)

t = (numerator%denominator , numerator//denominator)
print(t)
```

```python
#9. find the number of days between two dates
from datetime import date

x1 = int(input("enter year:"))
y1 = int(input("enter month:"))
z1 = int(input("enter day:"))
x2 = int(input("enter other persons year:"))
y2 = int(input("enter other persons month:"))
z2 = int(input("enter other persons day:"))

t1 = date(x1 , y1 , z1)
t2 = date(x2 , y2 , z2)

print(abs((t2-t1).days))

```
