```python
#accept three sides 
a = float(input("enter first side"))
b = float(input("enter second side"))
c = float(input("enter third side"))

#find the largest side

if a>b and a>c:
    print("hypotenuse is: a")
elif b>a and b>c:
    print("hypotenuse is: b")
else:
    print("hypotenuse is: c")
```
